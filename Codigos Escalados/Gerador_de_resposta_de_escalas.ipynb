```python
import pandas as pd
import random
from datetime import datetime, timedelta

# 1. Ler o arquivo original
df_original = pd.read_csv('RespostaEscala.csv')

# 2. CORREÇÃO: Filtrar e remover colunas vazias ou que o pandas chame de 'Unnamed' (por causa de vírgulas no final das linhas)
df_original = df_original.loc[:, ~df_original.columns.str.contains('^Unnamed')]

# 3. Limpar possíveis espaços nas pontas dos nomes das colunas
headers = [str(h).strip() for h in df_original.columns.tolist() if pd.notna(h)]

opcoes_resposta = ['Gostaria', 'Não Gostaria', 'Indiferente']
novas_linhas = []
base_time = datetime(2026, 1, 20, 8, 0, 0)

for j in range(1, 6):
  novas_linhas = []
  # 4. Gerar 35 registros aleatórios
  for i in range(1, 55):
      linha = []
      for col in headers:
          if col == 'Carimbo de data/hora':
              timestamp = base_time + timedelta(minutes=random.randint(1, 2880))
              linha.append(timestamp.strftime('%d/%m/%Y %H:%M:%S'))
          elif col == 'Endereço de e-mail':
              linha.append(f'funcionaria{i}@icmc.usp.br')
          elif col == 'Nome':
              linha.append(f'Funcionária {i}')
          else:
              linha.append(random.choice(opcoes_resposta))

      novas_linhas.append(linha)

  # 5. Criar o novo DataFrame
  df_novo = pd.DataFrame(novas_linhas, columns=headers)
  nome_arquivo = f'RespostasEscala_54_funcionarias_{j}.csv'


  # 6. Salvar o resultado final, sem o problema do unnamed
  df_novo.to_csv(nome_arquivo, index=False, encoding='utf-8')

  print("Arquivo gerado com sucesso!")
```
